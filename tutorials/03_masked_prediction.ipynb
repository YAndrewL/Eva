# 🧩 Masked reconstruction with Eva

This notebook contains how to use Eva to predict masked tokens and cross-modality prediction.

Depending on what you mask, the same model tells you what is
inside a hidden region, what an unmeasured marker looks like, or what the matching H&E looks like.
All three are below.

## Step 1. Set up

`patchify` / `unpatchify` go between an image and its token grid, `masking` builds the binary masks,
and `rearrange` (einops) reshapes tokens between the two layouts. They are only here to draw the
masked input, the model itself takes the whole image plus a mask.

```python
import os

# run from the repository root so that relative paths (config.yaml, examples/) resolve
if os.path.basename(os.getcwd()) == "tutorials":
    os.chdir("..")

import matplotlib.pyplot as plt
import numpy as np
import torch
from omegaconf import OmegaConf

from Eva.utils import load_from_hf
from utils.helpers import patchify, unpatchify, masking, rearrange
```

## Step 2. Load the pre-trained model

The default checkpoint `Eva_model.ckpt` is the pre-trained model, which is what reconstruction runs
on. `patch_size = 8` and `image_size = 224` come from `config.yaml`, so each channel of a patch
turns into a 28 x 28 = 784 token grid.

```python
conf = OmegaConf.load("config.yaml")
device = "cuda" if torch.cuda.is_available() else "cpu"
patch_size = 8
image_size = 224

model = load_from_hf(
    repo_id="yandrewl/Eva",
    conf=conf,
    device=device
)
```

## Step 3. Load an example patch

`examples/` holds one 224 x 224 MIF patch with 52 channels plus its matching H&E. `biomarkers` is
the panel in channel order, channels are always addressed by name, and `channel_selected` picks the
three markers used for the RGB figures below.

```python
# Load examples of H&E and MIF patches with size of 224*224
mif_patch_np = np.load("examples/MIF_patch.npy")
he_patch_np = np.load("examples/HE_patch.npy") / 255.0
biomarkers = np.load("examples/biomarkers.npy", allow_pickle=True).tolist()

print(biomarkers)
channel_selected = [biomarkers.index(i) for i in ['DAPI', 'CD45', 'aSMA']]
```

## Step 4. Helpers for masking and reconstruction

Both helpers get reused by everything further down.

`generate_masking_input` only draws what the model was given: patchify the image, zero out the
tokens where the mask is 1 (or paint them white with `set_reverse=True`), stitch it back together.
It never touches the model.

`generate_recon_images` is the one that runs the model. It samples a mask with the strategy and
ratio you ask for, calls the model with that `infer_mask`, and hands back the reconstruction along
with the mask that produced it.

`infer_mask` is `[C, N]` over channels and tokens. A 1 means the model has to predict that position,
a 0 means it is allowed to see it.

```python
# generate a masked image to show input image
def generate_masking_input(input_data, infer_mask, set_reverse=False):
    input_patches = patchify(input_data.to(torch.float32), patch_size=patch_size)
    B, H, W, P1, P2, C = input_patches.shape
    input_patches = rearrange(input_patches, "B H W P1 P2 C -> B C (H W) (P1 P2)")  # [B, C, N, D]
    masked_input = input_patches.to(device) * (1 - infer_mask[None, ..., None]).repeat(1, 1, 1, P1 * P2).to(device)  # masked input patches
    if set_reverse:  # reverse the masked color as white
        masked_input += infer_mask[None, ..., None].repeat(1, 1, 1, P1 * P2).to(device)  # set masked areas to 1 (white)
    masked_input = rearrange(masked_input, "B C (H W) (P1 P2) -> B H W P1 P2 C", B=B, H=H, W=W, P1=P1, P2=P2, C=C)
    masked_input = unpatchify(masked_input)
    return masked_input[0].detach().cpu().numpy()

# recontruct by Eva
def generate_recon_images(input_data, mask_strategy, masking_ratio, marker_in, makrer_out=None, channel_specified=None):
    masker = masking(masking_ratio, mask_strategy)

    if channel_specified is not None:
        assert mask_strategy == 'specified'
        infer_mask = masker(torch.randn(16, input_data.shape[-1], 
                                        (image_size // patch_size) ** 2, 2), channel_specified).to(device)
    else:
        infer_mask = masker(
            torch.randn(16, input_data.shape[-1], (image_size // patch_size) ** 2, 2),
        ).to(device)

    # generate a patch mask for infer, with shape [B, C, N, D], actually B and D does not matter :)
    image_in = input_data.to(torch.float32)
    if len(image_in.shape) == 3:
        image_in = image_in.unsqueeze(0)
    image_in = image_in.to(device)

    with torch.no_grad():
        image_recon, _, _ = model(
            image_in,
            marker_in=[marker_in],
            marker_out=makrer_out,
            infer_mask=infer_mask,
            channel_mask=None,
        )
    return image_recon[0].detach().cpu().numpy(), infer_mask
```

## Step 5. Reconstruction under different masking strategies

The strategy sets the shape of the mask, not how much of the image it covers:

| strategy | what gets hidden | what it tests |
|---|---|---|
| `random` | random (channel, token) positions across the whole `[C, N]` plane | is the joint channel-space structure learned |
| `patch` | whole spatial tokens, in every channel at once | can a missing region be filled in from its surroundings |
| `specified` | whole channels, at every token (the ratio is ignored) | can one marker be predicted from the rest of the panel |

### Step 5.1 Random masking

75% of the (channel, token) positions go missing. Every channel keeps a different quarter of its
tokens, so filling one channel back in can lean on the other channels at the same location.

```python
# 1. random masking
mif_patch = torch.from_numpy(mif_patch_np).unsqueeze(0)[..., channel_selected]
image_recon, infer_mask = generate_recon_images(mif_patch, "random", 0.75, ['DAPI', 'CD45', 'aSMA'])
masked_input = generate_masking_input(mif_patch, infer_mask)

fig, ax = plt.subplots(1, 3, sharey=True)
ax[0].imshow(mif_patch_np[..., channel_selected].astype(np.float32))
ax[1].imshow(masked_input)
ax[2].imshow(image_recon)

ax[0].set_title("Original")
ax[1].set_title("Masked input")
ax[2].set_title("Reconstructed")

ax[1].sharey(ax[0])
ax[2].sharey(ax[0])
```

### Step 5.2 Patch masking

This time whole tokens are dropped in every channel at once, so nothing at those locations is
visible anywhere and the model has to work from the tissue around them. The masked tokens are drawn
white (`set_reverse=True`) to make the grid easy to see.

```python
# 2. patch masking
image_recon, infer_mask = generate_recon_images(mif_patch, "patch", 0.75, ['DAPI', 'CD45', 'aSMA'])
masked_input = generate_masking_input(mif_patch, infer_mask, set_reverse=True)  # for better visualization, here set masked patches as 

fig, ax = plt.subplots(1, 3, sharey=True)
ax[0].imshow(mif_patch_np[..., channel_selected].astype(np.float32))
ax[1].imshow(masked_input)
ax[2].imshow(image_recon)

ax[0].set_title("Original")
ax[1].set_title("Masked input")
ax[2].set_title("Reconstructed")

ax[1].sharey(ax[0])
ax[2].sharey(ax[0])
```

### Step 5.3 Channel masking

`specified` drops whole channels. Here that is channel 0, PD1, with the other 51 markers left
visible, so the model predicts one marker from the rest of the panel. The ratio does nothing in this
mode.

```python
# 3. channel masking
selected_channel = 0
mif_patch = torch.from_numpy(mif_patch_np).unsqueeze(0)
# here we use other channels to predict one channel, index=0 is PD1
image_recon, infer_mask = generate_recon_images(mif_patch, "specified", 0.75, marker_in=biomarkers, channel_specified=[selected_channel])  
masked_input = generate_masking_input(mif_patch, infer_mask)  

fig, ax = plt.subplots(1, 2)
ax[0].imshow(mif_patch_np[..., selected_channel].astype(np.float32))
ax[1].imshow(image_recon[..., selected_channel]) 

ax[0].set_title("Original")
ax[1].set_title("Predicted")
```

## Step 6. Image translation (MIF -> H&E)

Cross-modality prediction is the same thing with a hand-written mask instead of a sampled one.
`input_np` sticks three zero H&E channels onto the MIF channels, `marker_in` names all of them while
`marker_out` asks for the H&E ones only, and `infer_mask` sets the last three rows to 1 so those are
exactly the channels that get predicted. H&E was reversed during training, so the prediction is
shown as `1 - image_recon`.

```python
# 4. Image translation (MIF -> H&E)
he_placeholder = np.zeros_like(he_patch_np)
input_np = np.concatenate([mif_patch_np, he_placeholder], axis=-1)
input_t = torch.from_numpy(input_np).unsqueeze(0).float().to(device)

marker_in = biomarkers + ['HECHA1', 'HECHA2', 'HECHA3']
marker_out = [['HECHA1', 'HECHA2', 'HECHA3']]

num_patches = (input_np.shape[0] // patch_size) * (input_np.shape[1] // patch_size)
infer_mask = torch.zeros(len(marker_in), num_patches, device=device)
infer_mask[-3:, :] = 1.0

with torch.no_grad():
    image_recon, _, _ = model(
        input_t,
        marker_in=[marker_in],
        marker_out=marker_out,
        infer_mask=infer_mask,
        channel_mask=None,
    )

fig, ax = plt.subplots(1, 3, sharey=True)
ax[0].imshow(mif_patch_np[..., channel_selected].astype(np.float32))
ax[1].imshow(he_patch_np.astype(np.float32))
ax[2].imshow(1 - image_recon.detach().cpu().numpy()[0])  # H&E color was reversed during training

ax[0].set_title('MIF')
ax[1].set_title("Original H&E")
ax[2].set_title("Predicted H&E")

ax[1].sharey(ax[0])
ax[2].sharey(ax[0])
```
